# Question 1.(a) — Price-Elastic Flexible Consumer Formulation

## 1. Problem Description & Notation

We consider a single price-elastic residential prosumer over a 24-hour horizon $\mathcal{T} = \{1, \dots, 24\}$ with hourly resolution ($\Delta t = 1\text{ h}$). The consumer owns a flexible electric load and a rooftop photovoltaic (PV) system, and interacts with the external distribution network via hourly electricity imports and exports.

---

## 2. Daily Optimization Problem Formulation

The consumer aims at maximizing its total daily net utility (consumer surplus), defined as the gross consumption utility minus PV generation costs and net grid procurement costs:

$$
\begin{aligned}
\max_{L_t, PV_t, P_t^{\mathrm{imp}}, P_t^{\mathrm{exp}}} \quad & \sum_{t=1}^{24} \Big( u^L L_t - c^{\mathrm{PV}} PV_t - p_t^{\mathrm{imp}} P_t^{\mathrm{imp}} + p_t^{\mathrm{exp}} P_t^{\mathrm{exp}} \Big) \\
\text{s.t.} \quad & L_t + P_t^{\mathrm{exp}} = PV_t + P_t^{\mathrm{imp}} \qquad \forall t \in \mathcal{T} \\
& L^{\min} \le L_t \le L^{\max} \qquad \forall t \in \mathcal{T} \\
& 0 \le PV_t \le PV_t^{\max} \qquad \forall t \in \mathcal{T} \\
& P_t^{\mathrm{imp}} \ge 0, \quad P_t^{\mathrm{exp}} \ge 0 \qquad \forall t \in \mathcal{T}
\end{aligned}
$$

---

## 3. Properties of the Problem

* **Problem Class:** Linear Program (LP). Both the objective function and all equality and inequality constraints are linear functions of the decision variables.
* **Convexity:** Convex optimization problem (maximizing a concave/linear objective function over a convex polyhedral set, which is equivalent to minimizing the negative surplus).
* **Feasibility:** The problem is **always feasible** for any realistic non-negative parameters ($L^{\min} \le L^{\max}$, $PV_t^{\max} \ge 0$) because grid exchanges are unconstrained: for any chosen values $L_t \in [L^{\min}, L^{\max}]$ and $PV_t \in [0, PV_t^{\max}]$, the power balance is satisfied by setting $P_t^{\mathrm{imp}} = \max(0, L_t - PV_t) \ge 0$ and $P_t^{\mathrm{exp}} = \max(0, PV_t - L_t) \ge 0$.
* **Boundedness:** The problem is **bounded from above** if and only if $p_t^{\mathrm{exp}} \le p_t^{\mathrm{imp}}$ for all $t \in \mathcal{T}$. Since $\tau^{\mathrm{imp}} > 0$ and $\tau^{\mathrm{exp}} > 0$, we have:
  $$p_t^{\mathrm{imp}} - p_t^{\mathrm{exp}} = (p_t + \tau^{\mathrm{imp}}) - (p_t - \tau^{\mathrm{exp}}) = \tau^{\mathrm{imp}} + \tau^{\mathrm{exp}} > 0 \implies p_t^{\mathrm{exp}} < p_t^{\mathrm{imp}}$$
  This strictly prevents risk-free circular arbitrage (buying from the grid to resell simultaneously at a higher price), ensuring a finite optimal solution.
* **Problem Size:**
  * **Variables:** $4 \times 24 = 96$ continuous decision variables.
  * **Constraints:** $24$ equality balance constraints and $6 \times 24 = 144$ bound inequality constraints.

---

## 4. Hourly Decomposition and Techno-Economic Meaning

Because there are no intertemporal coupling constraints linking consecutive hours (e.g., no state-of-charge dynamics, no ramp rates, and no daily energy quotas), the daily objective function and constraint sets are fully additive and separable per hour $t$:

$$\max \sum_{t=1}^{24} \Big( u^L L_t - c^{\mathrm{PV}} PV_t - p_t^{\mathrm{imp}} P_t^{\mathrm{imp}} + p_t^{\mathrm{exp}} P_t^{\mathrm{exp}} \Big) = \sum_{t=1}^{24} \max \Big( u^L L_t - c^{\mathrm{PV}} PV_t - p_t^{\mathrm{imp}} P_t^{\mathrm{imp}} + p_t^{\mathrm{exp}} P_t^{\mathrm{exp}} \Big)$$

### The Independent Hourly Problem (for a given hour $t$)

$$
\begin{aligned}
\max_{L_t, PV_t, P_t^{\mathrm{imp}}, P_t^{\mathrm{exp}}} \quad & u^L L_t - c^{\mathrm{PV}} PV_t - p_t^{\mathrm{imp}} P_t^{\mathrm{imp}} + p_t^{\mathrm{exp}} P_t^{\mathrm{exp}} \\
\text{s.t.} \quad & L_t + P_t^{\mathrm{exp}} = PV_t + P_t^{\mathrm{imp}} \\
& L^{\min} \le L_t \le L^{\max} \\
& 0 \le PV_t \le PV_t^{\max} \\
& P_t^{\mathrm{imp}} \ge 0, \quad P_t^{\mathrm{exp}} \ge 0
\end{aligned}
$$

### Significance of the Decomposition

* **Computationally:** The daily 96-variable LP reduces to solving 24 independent, very small subproblems of 4 variables each. These can be solved in parallel, significantly reducing computational overhead and ensuring linear scalability with horizon length ($\mathcal{O}(24 \times 1)$ instead of polynomial growth).
* **Economically:** The consumer behaves as a **memoryless agent**. A consumption or generation decision made at hour $t$ has no operational or economic repercussions on any previous or future hours. Each hour represents an isolated, instantaneous trade-off between marginal consumption utility, PV generation cost, and prevailing effective grid tariffs.

# Question 1.(b) — Lagrangian, Dual Problem and KKT Conditions

For each hour $t$, the original maximization problem can equivalently be written as the following minimization problem:

$$
\begin{aligned}
\min_{L_t,PV_t,P_t^{\mathrm{imp}},P_t^{\mathrm{exp}}}
\quad
&-u^L L_t+c^{PV}PV_t
+p_t^{\mathrm{imp}}P_t^{\mathrm{imp}}
-p_t^{\mathrm{exp}}P_t^{\mathrm{exp}}
\end{aligned}
$$

subject to

$$
L_t+P_t^{\mathrm{exp}}-PV_t-P_t^{\mathrm{imp}}=0,
$$

$$
L^{\min}-L_t\leq0,\qquad
L_t-L^{\max}\leq0,
$$

$$
-PV_t\leq0,\qquad
PV_t-PV_t^{\max}\leq0,
$$

$$
-P_t^{\mathrm{imp}}\leq0,\qquad
-P_t^{\mathrm{exp}}\leq0.
$$

## Lagrangian

Let $\lambda_t\in\mathbb{R}$ be the multiplier associated with the power-balance equality. Let the inequality multipliers be

$$
\mu_t^{L,-},\mu_t^{L,+},
\mu_t^{PV,-},\mu_t^{PV,+},
\mu_t^{\mathrm{imp}},\mu_t^{\mathrm{exp}}\geq0.
$$

The Lagrangian is

$$
\begin{aligned}
\mathcal L_t
={}&-u^L L_t
+c^{PV}PV_t
+p_t^{\mathrm{imp}}P_t^{\mathrm{imp}}
-p_t^{\mathrm{exp}}P_t^{\mathrm{exp}}
\\
&+\lambda_t
(L_t+P_t^{\mathrm{exp}}-PV_t-P_t^{\mathrm{imp}})
\\
&+\mu_t^{L,-}(L^{\min}-L_t)
+\mu_t^{L,+}(L_t-L^{\max})
\\
&+\mu_t^{PV,-}(-PV_t)
+\mu_t^{PV,+}(PV_t-PV_t^{\max})
\\
&+\mu_t^{\mathrm{imp}}(-P_t^{\mathrm{imp}})
+\mu_t^{\mathrm{exp}}(-P_t^{\mathrm{exp}}).
\end{aligned}
$$

## KKT Conditions

The KKT conditions consist of stationarity, primal feasibility, dual feasibility, and complementary slackness.

### 1. Stationarity

$$
p_t^{\mathrm{imp}}-\lambda_t-\mu_t^{\mathrm{imp}}=0,
$$

$$
-p_t^{\mathrm{exp}}+\lambda_t-\mu_t^{\mathrm{exp}}=0,
$$

$$
-u^L+\lambda_t-\mu_t^{L,-}+\mu_t^{L,+}=0,
$$

$$
c^{PV}-\lambda_t-\mu_t^{PV,-}+\mu_t^{PV,+}=0.
$$

### 2. Primal feasibility

$$
L_t+P_t^{\mathrm{exp}}=PV_t+P_t^{\mathrm{imp}},
$$

$$
L^{\min}\leq L_t\leq L^{\max},
$$

$$
0\leq PV_t\leq PV_t^{\max},
$$

$$
P_t^{\mathrm{imp}}\geq0,\qquad P_t^{\mathrm{exp}}\geq0.
$$

### 3. Dual feasibility

$$
\mu_t^{L,-},\mu_t^{L,+},
\mu_t^{PV,-},\mu_t^{PV,+},
\mu_t^{\mathrm{imp}},\mu_t^{\mathrm{exp}}\geq0,
$$

while

$$
\lambda_t\in\mathbb{R}.
$$

### 4. Complementary slackness

$$
\mu_t^{L,-}(L^{\min}-L_t)=0,
$$

$$
\mu_t^{L,+}(L_t-L^{\max})=0,
$$

$$
\mu_t^{PV,-}PV_t=0,
$$

$$
\mu_t^{PV,+}(PV_t-PV_t^{\max})=0,
$$

$$
\mu_t^{\mathrm{imp}}P_t^{\mathrm{imp}}=0,
$$

$$
\mu_t^{\mathrm{exp}}P_t^{\mathrm{exp}}=0.
$$

## Dual Problem

The Lagrange dual function is

$$
q_t(\lambda_t,\mu_t)
=
\inf_{L_t,PV_t,P_t^{\mathrm{imp}},P_t^{\mathrm{exp}}}
\mathcal L_t.
$$

Hence, the dual problem is

$$
\max_{\lambda_t,\mu_t} q_t(\lambda_t,\mu_t)
$$

subject to

$$
\mu_t^{L,-},\mu_t^{L,+},
\mu_t^{PV,-},\mu_t^{PV,+},
\mu_t^{\mathrm{imp}},\mu_t^{\mathrm{exp}}\geq0
$$

and the stationarity equations above.

Equivalently, the dual objective can be written as

$$
\max
\quad
\mu_t^{L,-}L^{\min}
-\mu_t^{L,+}L^{\max}
-\mu_t^{PV,+}PV_t^{\max}.
$$

For the full 24-hour problem, the dual objective is summed over all hours.

## Strong Duality and KKT Optimality

The hourly problem is a feasible and bounded linear program. Therefore, strong duality holds and the primal and dual optimal objective values are equal:

$$
p_t^*=d_t^*.
$$

The KKT conditions are both necessary and sufficient for optimality. Therefore, a primal-dual point is optimal if and only if it satisfies stationarity, primal feasibility, dual feasibility, and complementary slackness.

## Implication for Import and Export

From stationarity and complementary slackness,

$$
P_t^{\mathrm{imp}}>0
\quad\Rightarrow\quad
\mu_t^{\mathrm{imp}}=0
\quad\Rightarrow\quad
\lambda_t=p_t^{\mathrm{imp}},
$$

while

$$
P_t^{\mathrm{exp}}>0
\quad\Rightarrow\quad
\mu_t^{\mathrm{exp}}=0
\quad\Rightarrow\quad
\lambda_t=p_t^{\mathrm{exp}}.
$$

Since

$$
p_t^{\mathrm{imp}}-p_t^{\mathrm{exp}}
=
\tau^{\mathrm{imp}}+\tau^{\mathrm{exp}}>0,
$$

import and export cannot both be strictly positive at the optimum.


# Question 1.(c) — Interpretation of Dual Variables and Uniqueness

The dual variables provide an economic interpretation of the constraints in the hourly optimization problem. In particular, they measure the marginal value of relaxing the corresponding constraints. By complementary slackness, a non-binding inequality constraint has a zero multiplier, while a binding constraint may have a non-zero multiplier.

## 1. Power-balance dual variable

The hourly power-balance constraint is

$$
L_t + P_t^{\mathrm{exp}} - PV_t - P_t^{\mathrm{imp}} = 0,
\qquad \forall t\in\mathcal{T},
$$

with associated dual variable

$$
\lambda_t\in\mathbb{R}.
$$

The multiplier $\lambda_t$ represents the marginal value of the power-balance constraint in hour $t$. Economically, it measures the marginal change in the optimal objective value associated with a marginal change in the energy that must be balanced in that hour. Its unit is DKK/kWh. Since the power-balance constraint is an equality, $\lambda_t$ is unrestricted in sign.

Thus, $\lambda_t$ can be interpreted as the hourly marginal value of energy balance in the optimization problem.

## 2. PV upper-bound dual variable

The PV production is constrained by

$$
0\leq PV_t\leq PV_t^{\max}.
$$

The upper bound

$$
PV_t-PV_t^{\max}\leq0
$$

has the dual variable

$$
\mu_t^{PV,+}\geq0.
$$

The corresponding complementary-slackness condition is

$$
\mu_t^{PV,+}\left(PV_t-PV_t^{\max}\right)=0.
$$

Therefore, if

$$
PV_t<PV_t^{\max},
$$

the upper bound is non-binding and

$$
\boxed{\mu_t^{PV,+}=0}.
$$

In this case, increasing the available PV production by a small amount would not improve the optimal objective at the margin.

If instead

$$
PV_t=PV_t^{\max},
$$

the PV upper bound is binding and $\mu_t^{PV,+}$ may be positive. Its value measures the marginal benefit of increasing the available PV production limit $PV_t^{\max}$ by one additional unit.

Hence, a positive $\mu_t^{PV,+}$ indicates that PV availability is constraining the optimal solution in hour $t$.

## 3. Load-bound dual variables

The flexible load is subject to

$$
L^{\min}\leq L_t\leq L^{\max}.
$$

The lower and upper bounds have the dual variables

$$
\mu_t^{L,-}\geq0,
\qquad
\mu_t^{L,+}\geq0.
$$

Their complementary-slackness conditions are

$$
\mu_t^{L,-}(L^{\min}-L_t)=0,
$$

and

$$
\mu_t^{L,+}(L_t-L^{\max})=0.
$$

### Upper load bound

If

$$
L_t<L^{\max},
$$

the upper load constraint is non-binding, and therefore

$$
\boxed{\mu_t^{L,+}=0}.
$$

If

$$
L_t=L^{\max},
$$

the upper bound is binding and $\mu_t^{L,+}$ may be positive. In this case, $\mu_t^{L,+}$ measures the marginal improvement in the optimal objective that would result from increasing $L^{\max}$ by one unit.

### Lower load bound

Similarly, if

$$
L_t>L^{\min},
$$

then the lower bound is non-binding and

$$
\boxed{\mu_t^{L,-}=0}.
$$

If

$$
L_t=L^{\min},
$$

the lower bound is binding and $\mu_t^{L,-}$ may be positive. Its magnitude measures the marginal effect of relaxing the minimum-load requirement.

Therefore, the two load-bound multipliers quantify the marginal economic value of the flexibility limits on consumption.

## 4. Lower PV bound and import/export bounds

The lower PV bound

$$
PV_t\geq0
$$

has multiplier

$$
\mu_t^{PV,-}\geq0,
$$

with complementary slackness

$$
\mu_t^{PV,-}PV_t=0.
$$

Hence, if

$$
PV_t>0,
$$

then

$$
\mu_t^{PV,-}=0.
$$

Likewise, the non-negativity constraints on grid imports and exports,

$$
P_t^{\mathrm{imp}}\geq0,
\qquad
P_t^{\mathrm{exp}}\geq0,
$$

have multipliers

$$
\mu_t^{\mathrm{imp}}\geq0,
\qquad
\mu_t^{\mathrm{exp}}\geq0,
$$

with

$$
\mu_t^{\mathrm{imp}}P_t^{\mathrm{imp}}=0,
$$

$$
\mu_t^{\mathrm{exp}}P_t^{\mathrm{exp}}=0.
$$

Thus, whenever the corresponding import or export variable is strictly positive, its lower-bound multiplier must be zero.

## 5. Economic interpretation of the multipliers

The dual variables can therefore be interpreted as shadow prices of the corresponding constraints.

In general,

$$
\boxed{
\text{non-binding constraint}
\;\Rightarrow\;
\text{zero multiplier}
}
$$

while

$$
\boxed{
\text{binding constraint}
\;\Rightarrow\;
\text{multiplier may be positive}
}.
$$

A positive multiplier indicates that the corresponding constraint has a marginal economic value: relaxing the constraint slightly can improve the optimal objective.

For example:

- $\lambda_t$ measures the marginal value of the hourly power balance;
- $\mu_t^{PV,+}$ measures the marginal value of increasing the available PV limit $PV_t^{\max}$;
- $\mu_t^{L,+}$ measures the marginal value of increasing the maximum load $L^{\max}$;
- $\mu_t^{L,-}$ measures the marginal effect associated with relaxing the minimum load $L^{\min}$.

## 6. Uniqueness of the dual solution

Strong duality guarantees that the optimal primal and dual objective values are equal, but it does not guarantee that the optimal dual variables are unique.

Therefore,

$$
\boxed{
\text{strong duality}
\not\Rightarrow
\text{unique dual solution}.
}
$$

In a linear program, multiple sets of dual multipliers can satisfy the KKT conditions and attain the same optimal dual objective, particularly in degenerate cases where several constraints are simultaneously active.

Consequently,

$$
\boxed{
\text{unique primal solution}
\;\not\Rightarrow\;
\text{unique dual solution}.
}
$$

The primal variables describe the optimal operating point, whereas the dual variables describe the marginal value of the constraints at that point.


# Question 1.(d) — Price Ladder

## (i) No simultaneous import and export

The KKT conditions from Question 1.(b) can be used to establish that importing and exporting electricity cannot both be strictly positive at the optimum.

From the complementary-slackness conditions,

$$
\mu_t^{\mathrm{imp}}P_t^{\mathrm{imp}}=0,
$$

$$
\mu_t^{\mathrm{exp}}P_t^{\mathrm{exp}}=0.
$$

If

$$
P_t^{\mathrm{imp}}>0,
$$

then

$$
\mu_t^{\mathrm{imp}}=0.
$$

Using the stationarity condition for $P_t^{\mathrm{imp}}$,

$$
 p_t^{\mathrm{imp}}-\lambda_t-\mu_t^{\mathrm{imp}}=0,
$$

we obtain

$$
\boxed{\lambda_t=p_t^{\mathrm{imp}}}.
$$

Similarly, if

$$
P_t^{\mathrm{exp}}>0,
$$

then

$$
\mu_t^{\mathrm{exp}}=0,
$$

and the stationarity condition

$$
-p_t^{\mathrm{exp}}+\lambda_t-\mu_t^{\mathrm{exp}}=0
$$

gives

$$
\boxed{\lambda_t=p_t^{\mathrm{exp}}}.
$$

Suppose, for contradiction, that both import and export are strictly positive:

$$
P_t^{\mathrm{imp}}>0,
\qquad
P_t^{\mathrm{exp}}>0.
$$

Then both relations must hold simultaneously, implying

$$
 p_t^{\mathrm{imp}}=p_t^{\mathrm{exp}}.
$$

However, from the effective import and export prices defined in Question 1.(a),

$$
 p_t^{\mathrm{imp}}=p_t+\tau^{\mathrm{imp}},
$$

$$
 p_t^{\mathrm{exp}}=p_t-\tau^{\mathrm{exp}},
$$

so that

$$
 p_t^{\mathrm{imp}}-p_t^{\mathrm{exp}}
 =\tau^{\mathrm{imp}}+\tau^{\mathrm{exp}}>0.
$$

Hence,

$$
 p_t^{\mathrm{imp}}>p_t^{\mathrm{exp}},
$$

which contradicts the equality required by simultaneous positive import and export.

Therefore,

$$
\boxed{
P_t^{\mathrm{imp}}P_t^{\mathrm{exp}}=0
}
$$

at the optimum. In other words, the prosumer either imports electricity from the grid or exports electricity to the grid in a given hour, but does not do both simultaneously.

This result is the starting point for the price-ladder analysis below.

# Question 2.(a) — Separable Convex Disutility Formulation

& 0 \le PV_t \le PV_t^{\max} \qquad \forall t \in \mathcal{T} \\
& P_t^{\mathrm{imp}} \ge 0, \quad P_t^{\mathrm{exp}} \ge 0 \qquad \forall t \in \mathcal{T}
\end{aligned}
$$

  * *Question 2.(a):* **Nonlinear Convex Optimization Problem** (NLP) due to the presence of the general nonlinear disutility function $D_t$ in the objective.
* **Convexity:** 
  * Preserved. Since $D_t(L_t, \ell^{\mathrm{ref}}_t)$ is assumed to be convex and all remaining procurement terms and constraints are linear affine, the objective function is convex (for minimization) over a polyhedral convex feasible region.
* **Feasibility & Boundedness:** 
  * Identical to Question 1.(a). Because the constraint set is unchanged and grid exchanges remain unrestricted, the problem is **always feasible**. It remains **bounded** from above because $D_t \ge 0$ and $p_t^{\mathrm{imp}} > p_t^{\mathrm{exp}}$.
* **Hourly Decomposability:** 
  * Preserved. Because disutility is explicitly assumed to be **separable per hour** (i.e., $D_t$ depends solely on instantaneous consumption $L_t$), there are no coupling constraints linking consecutive time steps. The 24-hour problem decomposes into 24 independent hourly subproblems:
  $$
  \max_{L_t, PV_t, P_t^{\mathrm{imp}}, P_t^{\mathrm{exp}}} \Big( - D_t(L_t, \ell^{\mathrm{ref}}_t) - c^{\mathrm{PV}} PV_t - p_t^{\mathrm{imp}} P_t^{\mathrm{imp}} + p_t^{\mathrm{exp}} P_t^{\mathrm{exp}} \Big) \quad \text{s.t. hourly constraints}
  $$

## 4. Question 1 as a Special Case of Disutility

The linear marginal utility objective of Question 1 ($\max \sum_t u^L L_t$) corresponds to a specific choice of reference profile and disutility function:

### 1. Choice of Reference Profile and Disutility Function
* Set the reference profile to the upper consumption bound:
  $$\ell^{\mathrm{ref}}_t = L^{\max} \quad \forall t \in \mathcal{T}$$
* Define the disutility function as a linear function of the deviation from this upper limit:
  $$D_t(L_t, \ell^{\mathrm{ref}}_t) = u^L \cdot (L^{\max} - L_t)$$

### 2. Demonstration of Equivalence
Substituting this choice into the objective function yields:
$$
\begin{aligned}
\max \sum_{t=1}^{24} \Big( - u^L (L^{\max} - L_t) - \text{costs} \Big) &= \max \sum_{t=1}^{24} \Big( u^L L_t - u^L L^{\max} - \text{costs} \Big) \\
&= \max \sum_{t=1}^{24} \Big( u^L L_t - \text{costs} \Big) - \underbrace{\sum_{t=1}^{24} u^L L^{\max}}_{\text{constant}}
\end{aligned}
$$
Because subtracting a constant term does not affect the argmax, the set of optimal decision variables is identical to that of Question 1.(a).

### 3. Assumption Relaxed
This formulation relaxes the assumption of **bilateral penalty / local satiation around an interior reference point**:
* In the general disutility model of Question 2, any deviation either above or below $\ell^{\mathrm{ref}}_t$ causes discomfort ($D_t \ge 0$ with an interior minimum at $\ell^{\mathrm{ref}}_t$).
* In Question 1, consumption is **strictly monotonic (non-satiated)** across the entire operational domain $[L^{\min}, L^{\max}]$: consuming more electricity always yields higher utility without penalizing high consumption.
